```python
updated = message_pass(features, neighbors)
assert len(updated) == len(features)
# Node 4 has a single neighbor, node 3, so its new feature is node 3's feature.
assert updated[4] == features[3], "Check that each node reads its neighbors' current features."
# In an undirected graph, each node's feature is counted once for every neighbor it has.
assert sum(updated) == sum(len(neighbors[v]) * features[v] for v in neighbors)
print("Sum-aggregation checks passed:", updated)
```

```python
def aggregate_variant(features, neighbors, node, mode="sum", include_self=False):
    values = [features[nbr] for nbr in neighbors[node]]
    if include_self:
        values.append(features[node])
    if mode == "sum":
        return sum(values)
    if mode == "mean":
        return sum(values) / len(values)
    raise ValueError("mode must be 'sum' or 'mean'")

node = 2
print({"sum": aggregate_variant(features, neighbors, node, "sum"),
       "mean": round(aggregate_variant(features, neighbors, node, "mean"), 3),
       "sum_with_self": aggregate_variant(features, neighbors, node, "sum", include_self=True)})
```
