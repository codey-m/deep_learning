# Coding practice: stable and masked autoregressive NLL

An autoregressive model gives each step of a sequence a conditional probability. Turn those into the sequence's negative log-likelihood, then into a per-token figure for a padded batch that counts only real target tokens.

It is the sum of `-log p` over the per-step probabilities, which equals the negative log of their product without ever forming that product.

```python
import math

conditional_probs = [0.5, 0.25, 0.5]
padded_batch = [
    ([0.8, 0.5, 0.25, 1.0], [1, 1, 1, 0]),
    ([0.6, 0.5, 1.0, 1.0], [1, 1, 0, 0]),
]
```

```python
def sequence_nll(probs):
    """Return the summed sequence NLL."""
    if any(p <= 0 or p > 1 for p in probs):
        raise ValueError("Each observed-token probability must lie in (0, 1].")
    # TODO: return the sequence NLL without forming the product of the probabilities.
    raise NotImplementedError("Return the sequence NLL.")

def masked_token_nll(batch):
    """Return total NLL divided by the number of valid target tokens."""
    total = 0.0
    valid = 0
    for probs, mask in batch:
        for p, keep in zip(probs, mask):
            # TODO: accumulate the loss and the count over positions that hold a real target.
            raise NotImplementedError("Accumulate the loss and count over real targets.")
    if valid == 0:
        raise ValueError("The batch has no valid target tokens.")
    return total / valid
```

```python
# With every position valid, the masked mean is the sequence NLL over its length.
assert abs(sequence_nll(conditional_probs) / 3 - masked_token_nll([(conditional_probs, [1, 1, 1])])) < 1e-12
print("sequence summed NLL =", round(sequence_nll(conditional_probs), 4))
print("batch mean NLL per valid token =", round(masked_token_nll(padded_batch), 4))
```
